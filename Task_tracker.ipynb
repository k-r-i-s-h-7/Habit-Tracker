```python
from langchain.prompts import ChatPromptTemplate

from langchain_groq import ChatGroq
from dotenv import load_dotenv
import os
load_dotenv()

os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
llm = ChatGroq(model="deepseek-r1-distill-llama-70b",temperature=0)
```

```python
import random
from langchain.tools import tool
@tool
def motivation():
    """Will give a link to motivational video """
    motivational_videos = [
        "https://www.youtube.com/watch?v=teVXlKRZ8Dg",
        "https://www.youtube.com/watch?v=lFjs69xipFs",
        "https://www.youtube.com/watch?v=d96akWDnx0w",
        "https://www.youtube.com/watch?v=ZebSXPUCPFc",
        "https://www.youtube.com/watch?v=UNQhuFL6CWg",
        "https://www.youtube.com/watch?v=wnHW6o8WMas",
        "https://www.youtube.com/watch?v=mgmVOuLgFB0",
        "https://www.youtube.com/watch?v=mgmVOuLgFB0",
        "https://www.youtube.com/watch?v=IdTMDpizis8",
        "https://www.youtube.com/watch?v=2Lz0VOltZKA",
        "https://www.youtube.com/watch?v=26U_seo0a1g",
        "https://www.youtube.com/watch?v=ZXsQAXx_ao0",
        "https://www.youtube.com/watch?v=26U_seo0a1g",
        "https://www.youtube.com/watch?v=UfcAVejslrU",
        "https://www.youtube.com/watch?v=mgmVOuLgFB0",
        "https://www.youtube.com/watch?v=KxGRhd_iWuE",
        "https://www.youtube.com/watch?v=HZ1SXVdZPqw",
        "https://www.youtube.com/watch?v=btPJPFnesV4",
        "https://www.youtube.com/watch?v=wnHW6o8WMas",
        "https://www.youtube.com/watch?v=0wFBMZ4Qt2Q"
    ]
    random_video = random.choice(motivational_videos)
    return random_video

```

```python
from langchain.agents import StructuredChatAgent, AgentExecutor
tools = [motivation]
prompt = ChatPromptTemplate.from_messages([
    ("system", """You are a helpful assistant. The user has give the reason he has not performed a task along with the reason for the task"
        Based on the reason, you need to generate a response that is helpful and informative.
        Also use the tools to soothe the user's feelings and provide a solution to the problem and give a uplifting answer to the user
        Include the link in the final output"""),
    ("user", "{input}"),
])
agent = StructuredChatAgent.from_llm_and_tools(
    tools=tools,
    prompt=prompt,
    llm=llm
)
agent_ex = AgentExecutor(
    agent=agent,
    tools=tools,
    verbose=True,
    return_intermediate_steps=True,
)
```

```python
response = agent_ex.invoke("I wwas feeling bored")
```

```python
# First, add these imports to your existing imports
from google.oauth2.credentials import Credentials
from google_auth_oauthlib.flow import InstalledAppFlow
from googleapiclient.discovery import build
from datetime import datetime, timedelta
import pytz

# Add new tool functions
@tool
def guided_meditation():
    """Provides a link to a guided meditation video."""
    meditation_videos = [
        "https://www.youtube.com/watch?v=inpok4MKVLM",
        "https://www.youtube.com/watch?v=ZToicYcHIOU",
        "https://www.youtube.com/watch?v=O-6f5wQXSu8",
        "https://www.youtube.com/watch?v=SEfs5TJZ6Nk",
        "https://www.youtube.com/watch?v=5GSeWdjyr1c"
    ]
    return random.choice(meditation_videos)

@tool
def exercise_routine():
    """Provides a link to quick exercise routines."""
    exercise_videos = [
        "https://www.youtube.com/watch?v=UBMk30rjy0o",
        "https://www.youtube.com/watch?v=ml6cT4AZdqI",
        "https://www.youtube.com/watch?v=oAPCPjnU1wA"
    ]
    return random.choice(exercise_videos)

# Update your tools list
tools = [motivation, guided_meditation, exercise_routine]

# Update the system prompt
prompt = ChatPromptTemplate.from_messages([
    ("system", """You are a helpful assistant. When the user shares their mood or situation:
        1. Analyze their emotional state
        2. Suggest appropriate activities (meditation, exercise, or motivation)
        3. Ask if they would like to schedule any of these activities for tomorrow
        4. Use the schedule_activity tool if they agree to schedule
        5. Provide relevant video links using the tools
        Be empathetic and supportive in your response."""),
    ("user", "{input}"),
])

# Update agent and executor
agent = StructuredChatAgent.from_llm_and_tools(
    tools=tools,
    prompt=prompt,
    llm=llm
)

agent_ex = AgentExecutor(
    agent=agent,
    tools=tools,
    verbose=True,
    return_intermediate_steps=True,
)
```

```python
response  =agent_ex.invoke("I am feeling a bit down today. I need some motivation and maybe a quick workout to lift my spirits.")
```
